```python
import hashlib

texte = "Bonjour BTS SIO"
# 1. Encodage (String -> Bytes)
texte_bytes = texte.encode('utf-8')
# 2. Hachage
hash_object = hashlib.sha256(texte_bytes)
# 3. Conversion (Bytes -> Hex String)
empreinte = hash_object.hexdigest()

print(empreinte)
# Résultat : a2c4e... (64 caractères)
```

```python
class Block:
  def __init__(self, index:int, timestamp:str, data:str,
               previous_hash:str, hash:str):
    self.Index = index
    self.Timestamp = timestamp
    self.Data = data
    self.Previous_hash = previous_hash
    self.Hash = hash

  def calculate_hash(self):
    temp = self.Index + self.Timestamp + self.Data
    + self.Previous_hash + self.Hash
    return hashlib.sha256(temp.encode('utf-8')).hexdigest()

test = Block(2,'02/03/2026',"Alice doit 10€ à Bob","oieurgogikudfgdgdoigdiogdogfoi",)
```
